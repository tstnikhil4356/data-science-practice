```python
def speed_category(speed):
    if speed<=40:
        return "Slow"
    if speed > 40 & speed <=80:
        return "Medium"
    if speed > 80:
        return "Fast"
df["Speed_Category"] = df["Speed"].apply(speed_category) 

difrentiation = df["Speed_Category"].value_counts()
print(difrentiation)
```

```python
df["Attack-to-Defense-Ratio"] = df["Attack"]/df["Defense"]
```

```python
maxrat = df["Attack-to-Defense-Ratio"].max()
minrat = df["Attack-to-Defense-Ratio"].min()
avgrat = df["Attack-to-Defense-Ratio"].mean()
```

```python
print("Maximum ration",maxrat)
print("Minimum ration",minrat)
print("Average ration",avgrat)
```

```python
total_pokemon = len(df)
total_legendary = len(df["Legendary"] == True)
avg_gen = df.groupby("Generation")[["HP", "Attack", "Defense", "Speed"]].mean()
most_com_type = df["Type 1"].value_counts().max()
best_gen = df.groupby("Generation")[["Total Stats"]].mean().max()


print("=" * 50)
print("POKEMON DATASET SUMMARY REPORT")
print("=" * 50)
print(f"\nTotal Pokemon: {total_pokemon}")
print(f"Legendary Pokemon: {total_legendary}")
print(f"\nAverage Stats by Generation:\n{avg_gen}")
print(f"\nMost Common Type: {most_com_type}")
print(f"Generation with Highest Avg Total Stats: {best_gen}")
print("\n" + "=" * 50)
```
